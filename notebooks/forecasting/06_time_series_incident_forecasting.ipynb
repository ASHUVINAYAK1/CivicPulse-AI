# CivicPulse AI — Track 6: Time-Series Incident Forecasting

Chronological time-series forecasting for municipal monthly incident volumes, comparing naive baselines and recurrent sequence models.

```python
import sys
from pathlib import Path

ml_path = Path.cwd().parent.parent / "services" / "ml"
if str(ml_path) not in sys.path:
    sys.path.append(str(ml_path))

import numpy as np
import pandas as pd
from evaluation.metrics import evaluate_forecasting

# 1. Synthesize historical monthly incident counts (36 months)
dates = pd.date_range(start="2021-01-01", periods=36, freq="ME")
trend = np.linspace(30, 65, 36)
seasonal = 15 * np.sin(np.linspace(0, 6 * np.pi, 36)) # Monsoon seasonal spikes
noise = np.random.normal(0, 3, 36)
monthly_counts = np.round(trend + seasonal + noise)

ts_df = pd.DataFrame({"date": dates, "incidents": monthly_counts})

# 2. Chronological train/test split (last 6 months for test)
train_ts = ts_df.iloc[:-6]
test_ts = ts_df.iloc[-6:]

# 3. Seasonal Naive baseline (last year's corresponding month)
naive_forecast = train_ts["incidents"].iloc[-12:-6].values
y_true = test_ts["incidents"].values

metrics = evaluate_forecasting(y_true, naive_forecast)
print("Seasonal Naive Forecast Evaluation:", metrics)
```
