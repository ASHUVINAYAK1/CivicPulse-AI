# CivicPulse AI — Track 3: Unsupervised Learning (K-Means & PCA)

Discovers geographical damage hotspots and reduces feature dimensionality.

```python
import sys
from pathlib import Path

ml_path = Path.cwd().parent.parent / "services" / "ml"
if str(ml_path) not in sys.path:
    sys.path.append(str(ml_path))

import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from preprocessing.dataset_loader import generate_benchmark_synthetic_dataset
from evaluation.metrics import evaluate_clustering

# 1. Load dataset
df = generate_benchmark_synthetic_dataset(n_samples=400, random_state=42)
coords = df[["latitude", "longitude"]].values

# 2. K-Means Hotspot Clustering (k=5)
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(coords)
metrics = evaluate_clustering(coords, cluster_labels, kmeans.cluster_centers_)
print("Hotspot Cluster Evaluation:", metrics)

# 3. PCA Dimensionality Reduction
pca = PCA(n_components=2)
reduced_coords = pca.fit_transform(coords)
print("PCA Explained Variance Ratio:", pca.explained_variance_ratio_)
```
