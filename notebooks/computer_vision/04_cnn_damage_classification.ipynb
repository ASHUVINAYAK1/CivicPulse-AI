Pipeline for preprocessing RDD2022 imagery and evaluating damage classification into D00 (longitudinal), D10 (transverse), D20 (alligator), and D40 (pothole).

```python
import sys
from pathlib import Path

ml_path = Path.cwd().parent.parent / "services" / "ml"
if str(ml_path) not in sys.path:
    sys.path.append(str(ml_path))

import numpy as np
from PIL import Image
from preprocessing.image_pipeline import ImagePreprocessor
from evaluation.metrics import evaluate_multiclass_classification

# 1. Initialize image pipeline
preprocessor = ImagePreprocessor(target_size=(224, 224))

# 2. Demonstrate tensor conversion with synthetic test canvas
test_img = Image.new("RGB", (640, 480), color=(128, 128, 128))
tensor = preprocessor.preprocess_image(test_img)
print(f"Normalized Tensor Shape: {tensor.shape} (C, H, W)")
print(f"Tensor Range: [{tensor.min():.2f}, {tensor.max():.2f}]")

# 3. Benchmark evaluation logic with test predictions
y_true = np.array([0, 1, 2, 3, 3, 2, 0, 1])
y_pred = np.array([0, 1, 2, 3, 3, 2, 0, 0])
metrics = evaluate_multiclass_classification(y_true, y_pred, labels=["D00", "D10", "D20", "D40"])
print("Classifier Validation Metrics:", metrics)
```
