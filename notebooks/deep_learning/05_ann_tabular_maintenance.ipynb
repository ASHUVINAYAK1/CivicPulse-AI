# CivicPulse AI — Track 5: Deep Learning Artificial Neural Networks (ANN)

Evaluates multi-layer feedforward neural networks (Dense 128 -> ReLU -> Dropout -> Dense 64 -> Output) for non-linear priority and duration estimation.

```python
import sys
from pathlib import Path

ml_path = Path.cwd().parent.parent / "services" / "ml"
if str(ml_path) not in sys.path:
    sys.path.append(str(ml_path))

import numpy as np
from sklearn.neural_network import MLPClassifier
from preprocessing.dataset_loader import generate_benchmark_synthetic_dataset, split_incident_dataset
from preprocessing.tabular_pipeline import StructuredIncidentTransformer
from evaluation.metrics import evaluate_binary_classification

df = generate_benchmark_synthetic_dataset(n_samples=500, random_state=42)
train_df, val_df, test_df = split_incident_dataset(df)

transformer = StructuredIncidentTransformer()
X_train = transformer.fit_transform(train_df)
X_test = transformer.transform(test_df)
y_train = train_df["is_urgent"].values
y_test = test_df["is_urgent"].values

# Multi-Layer Perceptron (128, 64, 32)
mlp = MLPClassifier(hidden_layer_sizes=(128, 64, 32), max_iter=300, random_state=42)
mlp.fit(X_train, y_train)
preds = mlp.predict(X_test)
print("ANN Multi-layer Perceptron Metrics:", evaluate_binary_classification(y_test, preds))
```
